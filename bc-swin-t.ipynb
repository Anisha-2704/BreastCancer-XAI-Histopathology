```python
!pip install -q timm albumentations
import os
import gc
import random
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from PIL import Image
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import (
    Dataset,
    DataLoader,
    WeightedRandomSampler
)

from torchvision.models import swin_t, Swin_T_Weights

import albumentations as A
from albumentations.pytorch import ToTensorV2

from sklearn.model_selection import (
    StratifiedGroupKFold
)

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

from sklearn.preprocessing import label_binarize

import matplotlib.pyplot as plt
import seaborn as sns

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print(DEVICE)

MANIFEST_DIR = (
    "/kaggle/input/datasets/hritishachoudhury/bc-xai-quality-filtering"
)

df = pd.read_csv(
    f"{MANIFEST_DIR}/filtered_manifest.csv"
)

print(df.shape)

label_map = {
    "LumA": 0,
    "LumB": 1,
    "Her2": 2,
    "Basal": 3
}

df["label"] = (
    df["subtype_clean"]
    .map(label_map)
)

print(df.label.value_counts())

DATASET_ROOTS = {

    "bc-xai-a2-2000-partial":
    "/kaggle/input/datasets/anishapanja/bc-xai-a2-2000-partial",

    "bc-xai-e2-2000":
    "/kaggle/input/datasets/anishapanja/bc-xai-e2-2000",

    "bc-xai-a8-patches-batch1":
    "/kaggle/input/datasets/anishapanja/bc-xai-a8-patches-batch1",

    "bc-xai-a8-patches-batch2":
    "/kaggle/input/datasets/anishapanja/bc-xai-a8-patches-batch2",

    "bc-xai-a8-patches-batch3":
    "/kaggle/input/datasets/anishapanja/bc-xai-a8-patches-batch3",

    "bc-xai-a8-patches-batch4":
    "/kaggle/input/datasets/anishapanja/bc-xai-a8-patches-batch4",

    "bc-xai-a8-patches-batch5":
    "/kaggle/input/datasets/anishapanja/bc-xai-a8-patches-batch5",

    "bc-xai-c8-patches-batch1":
    "/kaggle/input/datasets/anishapanja/bc-xai-c8-patches-batch1",

    "bc-xai-c8-patches-batch2":
    "/kaggle/input/datasets/anishapanja/bc-xai-c8-patches-batch2",

    "bc-xai-c8-patches-batch3":
    "/kaggle/input/datasets/anishapanja/bc-xai-c8-patches-batch3",

    "bc-xai-c8-patches-batch4":
    "/kaggle/input/datasets/anishapanja/bc-xai-c8-patches-batch4",

    "bc-xai-bh-patches":
    "/kaggle/input/datasets/hritishachoudhury/bc-xai-bh-patches",

    "bc-xai-d8-patches":
    "/kaggle/input/datasets/hritishachoudhury/bc-xai-d8-patches"
}


def get_path(row):

    root = DATASET_ROOTS[row["dataset"]]

    return os.path.join(
        root,
        row["relative_path"]
    )


# augmentation
train_tfms = A.Compose([

    A.RandomResizedCrop(
        size=(300, 300),
        scale=(0.8, 1.0)
    ),

    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.5),

    A.Rotate(limit=90, p=0.5),

    A.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2,
        hue=0.1,
        p=0.7
    ),

    A.RandomGamma(
        gamma_limit=(80, 120),
        p=0.4
    ),

    A.GaussNoise(p=0.3),

    A.GaussianBlur(p=0.2),

    A.Normalize(),

    ToTensorV2()
])

val_tfms = A.Compose([

    A.Resize(300, 300),

    A.Normalize(),

    ToTensorV2()
])


# datasets
class BCDataset(Dataset):

    def __init__(self, df, transforms):

        self.df = df
        self.transforms = transforms

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        img = Image.open(
            get_path(row)
        ).convert("RGB")

        img = np.array(img)

        img = self.transforms(
            image=img
        )["image"]

        label = int(row.label)

        patient = row.patient_id

        return img, label, patient


# focal loss
class FocalLoss(nn.Module):

    def __init__(self, gamma=2):
        super().__init__()

        self.gamma = gamma

    def forward(
        self,
        logits,
        targets
    ):

        ce = F.cross_entropy(
            logits,
            targets,
            reduction="none",
            label_smoothing=0.1
        )

        pt = torch.exp(-ce)

        loss = (
            ((1 - pt) ** self.gamma)
            * ce
        )

        return loss.mean()


# full 5-fold training
# NOTE: epochs reduced from 10 to 7 and T_max from 10 to 7 solely to fit
# 5-fold CV within Kaggle's 9-hour commit limit. At ~12 min/epoch this gives
# 5 x 7 x 12 = ~7 hours worst-case. EfficientNet results showed best val_f1
# at epoch 6, so 7 epochs preserves full learning signal while early stopping
# (patience=5) handles any fold that converges earlier. Scientific comparison
# with EfficientNet remains valid: same folds, same data, same protocol.
sgkf = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

all_results = []

for fold, (train_idx, val_idx) in enumerate(
    sgkf.split(
        df,
        df["label"],
        groups=df["patient_id"]
    )
):

    print("\n" + "=" * 60)
    print(f"FOLD {fold+1}/5")
    print("=" * 60)

    train_df = (
        df.iloc[train_idx]
        .reset_index(drop=True)
    )

    val_df = (
        df.iloc[val_idx]
        .reset_index(drop=True)
    )

    # fold-wise patient counts
    print(
        "Train Patients:",
        train_df.patient_id.nunique()
    )

    print(
        "Val Patients:",
        val_df.patient_id.nunique()
    )

    ###################################################
    # DATALOADER
    ###################################################

    counts = train_df.label.value_counts()

    weights = len(train_df) / counts

    sample_weights = (
        train_df.label.map(weights)
    )

    sampler = WeightedRandomSampler(
        sample_weights.values,
        len(sample_weights),
        replacement=True
    )

    train_ds = BCDataset(
        train_df,
        train_tfms
    )

    val_ds = BCDataset(
        val_df,
        val_tfms
    )

    BATCH_SIZE = 28

    train_loader = DataLoader(
        train_ds,
        batch_size=BATCH_SIZE,
        sampler=sampler,
        num_workers=4,
        pin_memory=True,
        persistent_workers=True,
        prefetch_factor=2
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=4,
        pin_memory=True,
        persistent_workers=True,
        prefetch_factor=2
    )

    ###################################################
    # MODEL
    ###################################################

    weights = Swin_T_Weights.DEFAULT
    model = swin_t(weights=weights)
    model.head = nn.Linear(model.head.in_features, 4)

    model = model.to(DEVICE)

    criterion = FocalLoss()

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=1e-4,
        weight_decay=5e-4
    )

    scheduler = (
        torch.optim.lr_scheduler
        .CosineAnnealingLR(
            optimizer,
            T_max=7
        )
    )

    scaler = torch.amp.GradScaler(enabled=(DEVICE == "cuda"))

    ###################################################
    # TRAINING
    ###################################################

    BEST = 0
    PATIENCE = 5
    counter = 0

    # store best val predictions during training to avoid
    # a redundant second inference pass after loading checkpoint
    best_preds = None
    best_targets = None
    best_avg_probs = None

    for epoch in range(7):

        print(f"\nEpoch {epoch+1}/7")

        model.train()

        losses = []

        for images, labels, _ in tqdm(train_loader):

            images = images.to(DEVICE)
            labels = labels.to(DEVICE)

            optimizer.zero_grad()

            with torch.amp.autocast(
                device_type=DEVICE,
                enabled=(DEVICE == "cuda")
            ):

                outputs = model(images)

                loss = criterion(
                    outputs,
                    labels
                )

            scaler.scale(loss).backward()

            scaler.step(optimizer)

            scaler.update()

            losses.append(loss.item())

        ###################################################
        # VALIDATION
        ###################################################

        model.eval()

        patient_probs = {}

        with torch.no_grad():

            for images, labels, patients in tqdm(val_loader):

                images = images.to(DEVICE)

                logits = model(images)

                probs = F.softmax(
                    logits,
                    dim=1
                )

                probs = probs.cpu().numpy()

                for p, prob in zip(
                    patients,
                    probs
                ):

                    if p not in patient_probs:
                        patient_probs[p] = []

                    patient_probs[p].append(prob)

        gt = (
            val_df
            .groupby("patient_id")
            .first()
        )

        preds = []
        targets = []
        avg_probs_epoch = []

        for patient in patient_probs:

            avg = np.mean(
                patient_probs[patient],
                axis=0
            )

            pred = np.argmax(avg)

            target = gt.loc[
                patient,
                "label"
            ]

            preds.append(pred)
            targets.append(target)
            avg_probs_epoch.append(avg)

        f1 = f1_score(
            targets,
            preds,
            average="macro"
        )

        print(
            "Loss:",
            np.mean(losses)
        )

        print(
            "Patient F1:",
            f1
        )

        scheduler.step()

        if f1 > BEST:

            BEST = f1

            counter = 0

            torch.save(
                model.state_dict(),
                f"best_swin_t_fold_{fold+1}.pth"
            )

            # cache the predictions from this best epoch to avoid
            # re-running inference after loading checkpoint
            best_preds = preds[:]
            best_targets = targets[:]
            best_avg_probs = np.array(avg_probs_epoch)

        else:
            counter += 1

        if counter >= PATIENCE:

            print("Early stopping")

            break

    ###################################################
    # FINAL EVALUATION
    # Use cached predictions from best epoch — avoids a full
    # second inference pass over the validation set, saving
    # ~1.2 min per fold x 5 folds = ~6 min total.
    # Results are identical to reloading checkpoint and re-running
    # inference because best_preds was captured at the same epoch
    # that produced best_swin_t_fold_{fold+1}.pth.
    ###################################################

    preds = best_preds
    targets = best_targets
    avg_probs = best_avg_probs

    acc = accuracy_score(
        targets,
        preds
    )

    precision = precision_score(
        targets,
        preds,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        targets,
        preds,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        targets,
        preds,
        average="macro",
        zero_division=0
    )

    try:

        auc = roc_auc_score(
            label_binarize(
                targets,
                classes=[0, 1, 2, 3]
            ),
            avg_probs,
            average="macro",
            multi_class="ovr"
        )

    except Exception as e:

        print(e)

        auc = np.nan

    print("\nFold Metrics")

    print(f"Accuracy : {acc:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"Macro F1 : {f1:.4f}")
    print(f"Macro AUC: {auc:.4f}")

    report = classification_report(
        targets,
        preds,
        target_names=[
            "LumA",
            "LumB",
            "Her2",
            "Basal"
        ],
        output_dict=True,
        zero_division=0
    )

    print(
        classification_report(
            targets,
            preds,
            target_names=[
                "LumA",
                "LumB",
                "Her2",
                "Basal"
            ],
            zero_division=0
        )
    )

    report_df = pd.DataFrame(report).T

    report_df.to_csv(
        f"classification_report_fold_{fold+1}.csv"
    )

    display(report_df)

    cm = confusion_matrix(
        targets,
        preds
    )

    plt.figure(figsize=(8, 6))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=[
            "LumA",
            "LumB",
            "Her2",
            "Basal"
        ],
        yticklabels=[
            "LumA",
            "LumB",
            "Her2",
            "Basal"
        ]
    )

    plt.title(
        f"Fold {fold+1} Confusion Matrix"
    )

    plt.savefig(
        f"cm_fold_{fold+1}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    pd.DataFrame({

        "patient_id": list(patient_probs.keys()),
        "true": targets,
        "pred": preds,

        "LumA_prob": avg_probs[:, 0],
        "LumB_prob": avg_probs[:, 1],
        "Her2_prob": avg_probs[:, 2],
        "Basal_prob": avg_probs[:, 3]

    }).to_csv(

        f"patient_predictions_fold_{fold+1}.csv",
        index=False
    )

    all_results.append({

        "fold": fold + 1,

        "accuracy": acc,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "auc": auc
    })

    # free GPU memory between folds
    del model, optimizer, scheduler, scaler
    del best_preds, best_targets, best_avg_probs
    gc.collect()
    torch.cuda.empty_cache()

###################################################
# CROSS-VALIDATION SUMMARY (outside fold loop)
###################################################

results = pd.DataFrame(all_results)

print("\nFINAL RESULTS")
display(results)

results = results.dropna(
    subset=["f1"]
)

best_fold = results.loc[
    results["f1"].idxmax()
]

print("\nBest Fold:")
print(best_fold)

best_fold_num = int(best_fold["fold"])

print(
    f"\nBest model is Fold {best_fold_num}"
)

pd.DataFrame(best_fold).T.to_csv(
    "best_fold_summary.csv",
    index=False
)

print(
    "\nMean Accuracy:",
    results.accuracy.mean(),
    "±",
    results.accuracy.std()
)

print(
    "Mean Precision:",
    results.precision.mean(),
    "±",
    results.precision.std()
)

print(
    "Mean Recall:",
    results.recall.mean(),
    "±",
    results.recall.std()
)

print(
    "Mean F1:",
    results.f1.mean(),
    "±",
    results.f1.std()
)

print(
    "Mean AUC:",
    results.auc.mean(),
    "±",
    results.auc.std()
)

results.to_csv(
    "cross_validation_results.csv",
    index=False
)
```
